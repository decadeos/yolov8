```python
import matplotlib.pyplot as plt

# конф = 0,25, перекрытие 0,2  каждой стороны

epochs = [10, 50, 150, 288]
d_p = [-0.0669, -0.0560, -0.0596, -0.1133]
d_r = [0.0066, 0.0022, 0.0022, 0.0155]

plt.figure(figsize=(8, 3))

plt.plot(epochs, d_p, marker='o', label='Delta Precision (d_p)', color='red')
plt.plot(epochs, d_r, marker='s', label='Delta Recall (d_r)', color='blue')

plt.title('Влияние SAHI на метрики в зависимости от эпох обучения')
plt.xlabel('Количество эпох')
plt.ylabel('Изменение метрики (SAHI - YOLO)')
plt.axhline(y=0, color='black', linestyle='--', linewidth=0.5)
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
